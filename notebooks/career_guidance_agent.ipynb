```python
from agents import Agent, Runner, trace
from openai.types.responses import ResponseTextDeltaEvent
```

```python
career_guidance_agent = Agent(
    name="Career Guidance Assistant",
    instructions="""
    You are a career guidance assistant.
    Help students choose suitable career paths.
    Explain the skills required for different jobs.
    Suggest courses and skills to learn.
    Provide career roadmaps for beginners.
    Give short, clear, and practical answers.
    """,
)
```

```python
with trace("Career Guidance Agent"):
    result = await Runner.run(
       career_guidance_agent ,
        "What skills should I learn to become an AI engineer?"
    )

print(result)
```

```python
response_stream = Runner.run_streamed(
    career_guidance_agent,
    "What skills should I learn to become an AI engineer?"
)

async for event in response_stream.stream_events():
    if event.type == "raw_response_event" and isinstance(
        event.data, ResponseTextDeltaEvent
    ):
        print(event.data.delta, end="", flush=True)
```
